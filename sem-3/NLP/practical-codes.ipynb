```python
#Experiment 1
import nltk
from nltk.parse import RecursiveDescentParser

# 1. Define a simple Context-Free Grammar (CFG)
grammar_definition = """
    S -> NP VP
    NP -> Det N | N
    VP -> V NP | V
    Det -> 'the' | 'a'
    N -> 'cat' | 'dog' | 'man' | 'telescope'
    V -> 'saw' | 'ate' | 'barked'
"""
cfg = nltk.CFG.fromstring(grammar_definition)

# 2. Initialize the parser
parser = RecursiveDescentParser(cfg)

# 3. Define a sentence (as a list of tokens) to parse
sentence = ['the', 'cat', 'saw', 'a', 'dog']
print(f"Grammar:\n{grammar_definition}")
print(f"Sentence to parse: {' '.join(sentence)}\n")

# 4. Parse the sentence and generate trees
try:
    print("Generated Parse Trees:")
    for tree in parser.parse(sentence):
        # Print the tree structure (can be visualized with tree.pretty_print() too)
        print(tree)
except ValueError as e:
    # NLTK can raise ValueError if parsing fails
    print(f"Parsing Failed: {e}")
```

```python
# Experiment 7
import numpy as np
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.naive_bayes import MultinomialNB
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score

# 1. Create a simple toy dataset (Spam vs. Ham)
messages = [
    "You won a free lunch! Call now.", # Spam
    "Meeting agenda sent. Please review.", # Ham
    "Urgent: Claim your prize money!", # Spam
    "Can we reschedule the dinner?", # Ham
    "Limited time offer! Free gift.", # Spam
    "Did you finish the assignment?", # Ham
]
# 0 = Ham, 1 = Spam
labels = np.array([1, 0, 1, 0, 1, 0])

# 2. Split data into training and testing sets
X_train, X_test, y_train, y_test = train_test_split(
    messages, labels, test_size=0.33, random_state=42
)
print(f"Training Samples: {len(X_train)}, Test Samples: {len(X_test)}")

# 3. Feature Extraction (Convert text to TF-IDF vectors)
tfidf_vectorizer = TfidfVectorizer(stop_words='english')
X_train_vectorized = tfidf_vectorizer.fit_transform(X_train)
X_test_vectorized = tfidf_vectorizer.transform(X_test)

# 4. Train the Multinomial Naïve Bayes model
nb_model = MultinomialNB()
nb_model.fit(X_train_vectorized, y_train)

# 5. Predict on the test set
y_pred = nb_model.predict(X_test_vectorized)

# 6. Evaluate the model
accuracy = accuracy_score(y_test, y_pred)
print("-" * 50)
print("Naïve Bayes Classifier Performance:")
print(f"True Labels (Test): {y_test}")
print(f"Predicted Labels:   {y_pred}")
print(f"Accuracy: {accuracy:.2f}")

# 7. Test on a new message
new_message = ["Congratulations, you have been selected for a trip!"]
new_message_vectorized = tfidf_vectorizer.transform(new_message)
prediction = nb_model.predict(new_message_vectorized)[0]
print(f"\nPrediction for '{new_message[0]}': {'Spam' if prediction == 1 else 'Ham'}")
```

```python
# Experiment 9
import numpy as np
from hmmlearn import hmm

# --- HMM Model for Weather (Hidden State) and Activity (Observation) ---
# States: S1=Rainy, S2=Sunny (Hidden)
# Observations: O1=Walk, O2=Shop, O3=Clean (Visible)

# 1. Define the model parameters manually (for simplicity)

# State-Transition Probabilities: P(St | St-1)
trans_prob = np.array([[0.7, 0.3], [0.4, 0.6]])

# Emission Probabilities: P(Ot | St)
emission_prob = np.array([[0.1, 0.6, 0.3], [0.5, 0.2, 0.3]])

# Initial State Probabilities: P(S1), P(S2)
start_prob = np.array([0.6, 0.4])

# Mapping for clarity
states_map = {0: "Rainy", 1: "Sunny"}
observations_map = {0: "Walk", 1: "Shop", 2: "Clean"}

# 2. Sequence of Observations (Activities over several days)
# Clean (2), Walk (0), Shop (1), Clean (2), Walk (0)
observations = np.array([[2, 0, 1, 2, 0]]).T  # Needs to be a column vector

# 3. Initialize the HMM
model = hmm.CategoricalHMM(n_components=len(states_map), random_state=42, n_iter=100)
model.startprob_ = start_prob
model.transmat_ = trans_prob
model.emissionprob_ = emission_prob

# 4. Predict the most probable sequence of hidden states (Weather)
# This uses the Viterbi algorithm internally
logprob, hidden_states = model.decode(observations, algorithm="viterbi")

print("HMM for Weather (Hidden) and Activity (Observed)")
print("-" * 50)
print(f"Observation Sequence: {[observations_map[o[0]] for o in observations]}")

# Map the predicted state indices back to names
predicted_states = [states_map[s] for s in hidden_states]

print(f"\nMost Probable Hidden State Sequence (Weather):")
print(f"{'Time Step':<10} | {'Activity':<10} | {'Predicted Weather':<20}")
print("-" * 50)
for i, (obs, state) in enumerate(zip(observations, predicted_states)):
    print(f"{i+1:<10} | {observations_map[obs[0]]:<10} | {state:<20}")
```

```python
# Experiment 10
import numpy as np

# --- Simplified Viterbi Implementation ---
# States: S1=H (High), S2=L (Low) - Hidden States
# Observations: O1=U (Up), O2=D (Down) - Visible Observations

# 1. Define Model Parameters
states = ('H', 'L')
observations = ('U', 'D', 'U', 'D') # Sequence
S_idx = {'H': 0, 'L': 1}
O_idx = {'U': 0, 'D': 1}

# Initial Probabilities (Pi)
start_prob = np.array([0.8, 0.2]) # P(H), P(L)

# Transition Matrix (A): A[i, j] = P(State j | State i)
trans_prob = np.array([[0.7, 0.3], [0.4, 0.6]])

# Emission Matrix (B): B[i, j] = P(Obs j | State i)
emission_prob = np.array([[0.9, 0.1], [0.2, 0.8]])

def viterbi(obs_seq, states, start_p, trans_p, emit_p):
    T = len(obs_seq) # Length of observation sequence
    N = len(states)  # Number of states
    
    # 1. Initialization: Create Viterbi and Backpointer tables
    V = np.zeros((N, T))
    path = np.zeros((N, T), dtype=int)
    
    # Time t=0 (First observation)
    for i in range(N):
        obs_idx = O_idx[obs_seq[0]]
        V[i, 0] = start_p[i] * emit_p[i, obs_idx]
    
    # 2. Recursion: For t > 0
    for t in range(1, T):
        obs_idx = O_idx[obs_seq[t]]
        for j in range(N): # Current state
            likelihoods = V[:, t-1] * trans_p[:, j] * emit_p[j, obs_idx]
            V[j, t] = np.max(likelihoods)
            path[j, t] = np.argmax(likelihoods)
    
    # 3. Termination and Backtracking
    most_likely_last_state = np.argmax(V[:, T-1])
    best_path = [most_likely_last_state]
    current_state = most_likely_last_state
    
    for t in range(T - 1, 0, -1):
        current_state = path[current_state, t]
        best_path.insert(0, current_state)
        
    result_path = [states[i] for i in best_path]
    max_prob = np.max(V[:, T-1])
    return result_path, max_prob, V

# 2. Run the Viterbi algorithm
predicted_sequence, probability, V_table = viterbi(
    observations, states, start_prob, trans_prob, emission_prob
)

# 3. Output results
print("Viterbi Algorithm Implementation (Custom)")
print("-" * 50)
print(f"Observations (Price Movement): {observations}")
print(f"Predicted Hidden States (Volatility): {predicted_sequence}")
print(f"Probability of Sequence: {probability:.4f}")
print("\nPartial Viterbi Table (V):")
print(V_table.round(5))
```

```python
# Experiment 11
import numpy as np
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.decomposition import TruncatedSVD

# 1. Sample Corpus (Documents)
corpus = [
    "Machine learning algorithms are used for classification.",
    "Natural language processing uses vectors to represent words.",
    "LSA is a technique for dimensionality reduction in NLP.",
    "Classification and regression are common ML tasks."
]
print("Original Corpus:")
for doc in corpus:
    print(f"- {doc}")
print("-" * 60)

# 2. Convert text to TF-IDF matrix
tfidf_vectorizer = TfidfVectorizer(stop_words='english')
tfidf_matrix = tfidf_vectorizer.fit_transform(corpus)
print(f"Original TF-IDF Matrix Shape: {tfidf_matrix.shape}")

# 3. Apply Latent Semantic Analysis (LSA) using Truncated SVD
# We want to reduce the dimension to K=2 semantic components (topics)
K = 2 
lsa_model = TruncatedSVD(n_components=K, random_state=42)
lsa_matrix = lsa_model.fit_transform(tfidf_matrix)

# 4. Output Results
print(f"\nReduced LSA Matrix Shape: {lsa_matrix.shape}")
print(f"LSA Matrix (Document-Topic scores, K={K}):")
print(lsa_matrix.round(3))
print("-" * 60)

# 5. Interpretation: Principal Components (Topics)
print("Interpretation (Top terms in each component/topic):")
feature_names = tfidf_vectorizer.get_feature_names_out()
for i, component in enumerate(lsa_model.components_):
    top_term_indices = component.argsort()[-3:][::-1]
    top_terms = [feature_names[j] for j in top_term_indices]
    print(f"  Topic {i+1}: {top_terms}")
```

```python
# Experiment 12
import gensim
from gensim import corpora
from nltk.tokenize import word_tokenize
import nltk

# Ensure NLTK data is available
try:
    nltk.data.find('tokenizers/punkt')
except:
    nltk.download('punkt', quiet=True)

# 1. Sample Corpus
documents = [
    "Federal reserve announces rate hike amidst inflation concerns.",
    "The new electric car has a long range and sleek design.",
    "Tesla introduces novel battery technology for vehicles.",
    "Central bank meeting discusses monetary policy and job growth.",
    "Investors monitor bond yields and commodity prices."
]

# 2. Preprocessing: Tokenization and creating a dictionary
tokenized_docs = [word_tokenize(doc.lower()) for doc in documents]
dictionary = corpora.Dictionary(tokenized_docs)
corpus_bow = [dictionary.doc2bow(doc) for doc in tokenized_docs]

print(f"Number of Unique Tokens: {len(dictionary)}")
print(f"Sample Document 1 (BoW):\n{corpus_bow[0]}\n")

# 3. Train the LDA Model
num_topics = 2 # We expect one topic for 'Finance' and one for 'Cars'
lda_model = gensim.models.LdaModel(
    corpus=corpus_bow,
    id2word=dictionary,
    num_topics=num_topics,
    random_state=42,
    passes=10 # Number of training passes
)

# 4. Output Results (Topics and associated words)
print("-" * 50)
print(f"LDA Topic Modeling (K={num_topics} Topics):")
topics = lda_model.print_topics(num_words=5)
for topic_id, word_dist in topics:
    print(f"Topic {topic_id + 1}: {word_dist}")

# 5. Document-Topic Distribution (Optional)
print("\nDocument Topic Distribution (Score for Topic 1, Topic 2):")
for i, doc_bow in enumerate(corpus_bow):
    doc_topics = lda_model.get_document_topics(doc_bow, minimum_probability=0.1)
    topic_scores = {topic_id: score for topic_id, score in doc_topics}
    score1 = topic_scores.get(0, 0.0)
    score2 = topic_scores.get(1, 0.0)
    print(f"Doc {i+1}: ({score1:.3f}, {score2:.3f})")
```

```python
# Experiment 13
import gensim
from gensim.models import Word2Vec
import nltk

# Ensure NLTK data is available
try:
    nltk.data.find('tokenizers/punkt')
except:
    nltk.download('punkt', quiet=True)

# 1. Sample Text Corpus
text = [
    "The quick brown fox jumps over the lazy dog.",
    "The dog barks at the fast cat.",
    "Dogs and cats are common pets.",
    "A quick animal is a fast animal."
]

# 2. Tokenize the corpus into sentences (list of lists of words)
tokenized_sentences = [nltk.word_tokenize(s.lower()) for s in text]

# 3. Train the Word2Vec Model (Skip-Gram, sg=1)
print("Training Word2Vec (Skip-Gram, vector_size=10)...")
model_sg = Word2Vec(
    sentences=tokenized_sentences,
    vector_size=10,
    window=2,
    min_count=1,
    sg=1, # Skip-Gram
    workers=4
)
print("Training complete.")

# 4. Explore the trained embeddings
word_vector = model_sg.wv['dog']
print(f"\nVector for 'dog' (first 5 elements):\n{word_vector[:5].round(4)}")

# Find the most similar words
print("\nWords most similar to 'cat':")
try:
    similar_words = model_sg.wv.most_similar('cat', topn=3)
    for word, similarity in similar_words:
        print(f"  - {word}: {similarity:.4f}")
except KeyError:
    print("Word 'cat' not found in vocabulary.")
```

```python
# Experiment 14
import gensim.downloader as api

# 1. Download a pre-trained GloVe model
print("Downloading GloVe-50 model. This may take a moment...")
try:
    glove_vectors = api.load("glove-wiki-gigaword-50")
    print("GloVe model loaded successfully.")
except Exception as e:
    print(f"Error loading model: {e}")
    exit()

# 2. Measure Similarity between words
word1 = "king"
word2 = "queen"
word3 = "man"

try:
    similarity_kq = glove_vectors.similarity(word1, word2)
    print("-" * 50)
    print("Word Similarity using GloVe Embeddings:")
    print(f"Similarity({word1}, {word2}): {similarity_kq:.4f}")

    # 3. Perform Word Analogies (e.g., King - Man + Woman = ?)
    print("\nWord Analogy: King - Man + Woman = ?")
    result = glove_vectors.most_similar(positive=['king', 'woman'], negative=['man'], topn=1)
    
    analogy_word, score = result[0]
    print(f"  - Result: {analogy_word} (Score: {score:.4f})")

    # 4. Find the word that is least similar
    print("\nOdd one out: [sun, moon, star, water]:")
    odd_one = glove_vectors.doesnt_match(['sun', 'moon', 'star', 'water'])
    print(f"  - Odd word: {odd_one}")

except KeyError as e:
    print(f"\nWord not found in GloVe vocabulary: {e}")
```

```python
# Experiment 15
from transformers import AutoTokenizer, AutoModelForSeq2SeqLM

# The T5 model is a simple, effective example of an encoder-decoder (Seq2Seq) model.
MODEL_NAME = "t5-small"

# 1. Load Tokenizer and Model
try:
    tokenizer = AutoTokenizer.from_pretrained(MODEL_NAME)
    model = AutoModelForSeq2SeqLM.from_pretrained(MODEL_NAME)
except Exception as e:
    print(f"Error loading {MODEL_NAME} model.")
    exit()

# 2. Define the input prompt with task prefix
input_text = "summarize: The COVID-19 pandemic significantly disrupted global supply chains. Factories closed, transportation costs soared, and consumer demand shifted rapidly towards digital goods. This led to widespread shortages and rising inflation across many sectors."

print(f"Input Task and Text:\n{input_text}\n")

# 3. Tokenize the input text
input_ids = tokenizer.encode(input_text, return_tensors="pt")

# 4. Generate the output sequence
output_ids = model.generate(
    input_ids,
    max_length=50,
    num_beams=4,
    early_stopping=True
)

# 5. Decode the generated tokens back to text
generated_text = tokenizer.decode(output_ids[0], skip_special_tokens=True)

# 6. Output the result
print("-" * 50)
print("Seq2Seq Generated Output (Summary):")
print(generated_text)
```

```python
# Experiment 18
import textdistance

# 1. Define a simple vocabulary (correct words)
VOCABULARY = set([
    "apple", "banana", "orange", "grape", "melon", 
    "computer", "science", "programming", "algorithm", "python"
])

def find_best_correction(misspelled_word, vocabulary):
    """
    Finds the correction in the vocabulary with the minimum
    Levenshtein distance (Edit Distance).
    """
    min_distance = float('inf')
    best_correction = None
    
    for correct_word in vocabulary:
        distance = textdistance.levenshtein.distance(misspelled_word, correct_word)
        
        if distance < min_distance:
            min_distance = distance
            best_correction = correct_word
        
    return best_correction, min_distance

# 2. Test cases
test_words = ["aple", "bannan", "programmin", "algoritum", "xyz"]

print("--- Spell Checker using Edit Distance (Levenshtein) ---")
print(f"Vocabulary Size: {len(VOCABULARY)}\n")
print(f"{'Word':<15} | {'Correction':<15} | {'Distance':<10}")
print("-" * 45)

# 3. Run the spell checker
for word in test_words:
    correction, distance = find_best_correction(word, VOCABULARY)
    
    if distance > 2: # Set a threshold for reasonable corrections
        correction = "No suggestion"
        distance = ""
    
    print(f"{word:<15} | {correction:<15} | {str(distance):<10}")
```

```python
# Experiment 19
from transformers import pipeline

# 1. Initialize the Sentiment Analysis pipeline
# We use a model fine-tuned for sentiment analysis
try:
    sentiment_pipeline = pipeline(
        "sentiment-analysis",
        model="distilbert-base-uncased-finetuned-sst-2-english"
    )
except Exception as e:
    print(f"Error loading model.")
    exit()

# 2. Define sample sentences
sentences = [
    "I absolutely loved the simplicity of this transformer code.",
    "The weather today is terrible and I feel sad.",
    "This is an incredibly neutral statement.",
    "The latency was a major drawback of the system.",
]

print("--- Sentiment Analysis using Pre-trained BERT (DistilBERT) ---")
print(f"Model used: distilbert-base-uncased-finetuned-sst-2-english\n")

# 3. Run analysis and output results
print(f"{'Sentence':<50} | {'Sentiment':<10} | {'Score':<5}")
print("-" * 70)

results = sentiment_pipeline(sentences)

for sentence, result in zip(sentences, results):
    label = result['label']
    score = result['score']
    
    print(f"{sentence[:48]+'...':<50} | {label:<10} | {score:.3f}")
```

```python
# Experiment 20
from transformers import pipeline

# 1. Initialize the Question Answering pipeline
# The pipeline uses a model fine-tuned on the SQuAD dataset
try:
    qa_pipeline = pipeline(
        "question-answering",
        model="distilbert-base-cased-distilled-squad"
    )
except Exception as e:
    print(f"Error loading model.")
    exit()

# 2. Define the context (paragraph) and the question
context = (
    "The invention of the printing press by Johannes Gutenberg around 1440 "
    "is widely regarded as the most important event of the modern period. "
    "It revolutionized information dissemination, making books much cheaper "
    "and more accessible to the general public. Before its invention, books "
    "had to be copied by hand, a slow and expensive process."
)
question = "Who invented the printing press and when?"

print("--- Question Answering (QA) using BERT (DistilBERT) ---")
print("-" * 70)
print(f"Context:\n{context}\n")
print(f"Question: {question}")
print("-" * 70)

# 3. Run the QA system
result = qa_pipeline(
    question=question,
    context=context
)

# 4. Output the result
answer = result['answer']
score = result['score']
start_index = result['start']
end_index = result['end']

print("QA Result:")
print(f"  Answer: {answer}")
print(f"  Confidence Score: {score:.4f}")
```
