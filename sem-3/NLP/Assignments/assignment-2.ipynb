```python
# EXperiment 2
import nltk
import spacy

# ---------------------------------------
# Download NLTK packages (safe try-except)
# ---------------------------------------

# punkt
try:
    nltk.data.find("tokenizers/punkt")
except LookupError:
    nltk.download("punkt")

# punkt_tab
try:
    nltk.data.find("tokenizers/punkt_tab")
except LookupError:
    nltk.download("punkt_tab")

# averaged_perceptron_tagger
try:
    nltk.data.find("taggers/averaged_perceptron_tagger")
except LookupError:
    nltk.download("averaged_perceptron_tagger")

# averaged_perceptron_tagger_eng (NEW)
try:
    nltk.data.find("taggers/averaged_perceptron_tagger_eng")
except LookupError:
    nltk.download("averaged_perceptron_tagger_eng")

# NE Chunker
try:
    nltk.data.find("chunkers/maxent_ne_chunker")
except LookupError:
    nltk.download("maxent_ne_chunker")

# Words corpus
try:
    nltk.data.find("corpora/words")
except LookupError:
    nltk.download("words")


# ---------------------------------------
# 1. Sample Text
# ---------------------------------------
text = "Google's stock closed at 145 dollars on Friday. It was a good day."
print(f"Original Text: {text}\n")

# ---------------------------------------
# 2. Tokenization
# ---------------------------------------
tokens = nltk.word_tokenize(text)
print(f"1. Tokenization:\n{tokens}\n")

# ---------------------------------------
# 3. POS Tagging
# ---------------------------------------
pos_tags = nltk.pos_tag(tokens)
print(f"2. POS Tagging:\n{pos_tags}\n")

# ---------------------------------------
# 4. Parsing (Regex Chunking)
# ---------------------------------------
grammar = r"""
  NP: {<DT|JJ|POS|CD>*<NN.*>+}
      {<NNP>+}
"""
cp = nltk.RegexpParser(grammar)
tree = cp.parse(pos_tags)
print(f"3. Simple Parsing (Chunking):\n{tree}\n")

# ---------------------------------------
# 5. Semantic/Entity Analysis (spaCy)
# ---------------------------------------
try:
    nlp = spacy.load("en_core_web_sm")
    doc = nlp(text)

    print("4. Semantic Analysis (Named Entities via spaCy):")
    for ent in doc.ents:
        print(f"  - Entity: {ent.text}, Type: {ent.label_}")

except OSError:
    print("4. Semantic Analysis (spaCy) FAILED:")
    print("   Please run: python -m spacy download en_core_web_sm")
```
