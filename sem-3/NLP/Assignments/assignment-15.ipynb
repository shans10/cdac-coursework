```python
# Experiment 15
from transformers import AutoTokenizer, AutoModelForSeq2SeqLM

# The T5 model is a simple, effective example of an encoder-decoder (Seq2Seq) model.
MODEL_NAME = "t5-small"

# 1. Load Tokenizer and Model
try:
    tokenizer = AutoTokenizer.from_pretrained(MODEL_NAME)
    model = AutoModelForSeq2SeqLM.from_pretrained(MODEL_NAME)
except Exception as e:
    print(f"Error loading {MODEL_NAME} model.")
    exit()

# 2. Define the input prompt with task prefix
input_text = "summarize: The COVID-19 pandemic significantly disrupted global supply chains. Factories closed, transportation costs soared, and consumer demand shifted rapidly towards digital goods. This led to widespread shortages and rising inflation across many sectors."

print(f"Input Task and Text:\n{input_text}\n")

# 3. Tokenize the input text
input_ids = tokenizer.encode(input_text, return_tensors="pt")

# 4. Generate the output sequence
output_ids = model.generate(
    input_ids,
    max_length=50,
    num_beams=4,
    early_stopping=True
)

# 5. Decode the generated tokens back to text
generated_text = tokenizer.decode(output_ids[0], skip_special_tokens=True)

# 6. Output the result
print("-" * 50)
print("Seq2Seq Generated Output (Summary):")
print(generated_text)
```
