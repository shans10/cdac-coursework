```python
#Experiment 1
import nltk
from nltk.parse import RecursiveDescentParser

# 1. Define a simple Context-Free Grammar (CFG)
grammar_definition = """
    S -> NP VP
    NP -> Det N | N
    VP -> V NP | V
    Det -> 'the' | 'a'
    N -> 'cat' | 'dog' | 'man' | 'telescope'
    V -> 'saw' | 'ate' | 'barked'
"""
cfg = nltk.CFG.fromstring(grammar_definition)

# 2. Initialize the parser
parser = RecursiveDescentParser(cfg)

# 3. Define a sentence (as a list of tokens) to parse
sentence = ['the', 'cat', 'saw', 'a', 'dog']
print(f"Grammar:\n{grammar_definition}")
print(f"Sentence to parse: {' '.join(sentence)}\n")

# 4. Parse the sentence and generate trees
try:
    print("Generated Parse Trees:")
    for tree in parser.parse(sentence):
        # Print the tree structure (can be visualized with tree.pretty_print() too)
        print(tree)
except ValueError as e:
    # NLTK can raise ValueError if parsing fails
    print(f"Parsing Failed: {e}")
```
