```python
# Experiment 11
import numpy as np
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.decomposition import TruncatedSVD

# 1. Sample Corpus (Documents)
corpus = [
    "Machine learning algorithms are used for classification.",
    "Natural language processing uses vectors to represent words.",
    "LSA is a technique for dimensionality reduction in NLP.",
    "Classification and regression are common ML tasks."
]
print("Original Corpus:")
for doc in corpus:
    print(f"- {doc}")
print("-" * 60)

# 2. Convert text to TF-IDF matrix
tfidf_vectorizer = TfidfVectorizer(stop_words='english')
tfidf_matrix = tfidf_vectorizer.fit_transform(corpus)
print(f"Original TF-IDF Matrix Shape: {tfidf_matrix.shape}")

# 3. Apply Latent Semantic Analysis (LSA) using Truncated SVD
# We want to reduce the dimension to K=2 semantic components (topics)
K = 2 
lsa_model = TruncatedSVD(n_components=K, random_state=42)
lsa_matrix = lsa_model.fit_transform(tfidf_matrix)

# 4. Output Results
print(f"\nReduced LSA Matrix Shape: {lsa_matrix.shape}")
print(f"LSA Matrix (Document-Topic scores, K={K}):")
print(lsa_matrix.round(3))
print("-" * 60)

# 5. Interpretation: Principal Components (Topics)
print("Interpretation (Top terms in each component/topic):")
feature_names = tfidf_vectorizer.get_feature_names_out()
for i, component in enumerate(lsa_model.components_):
    top_term_indices = component.argsort()[-3:][::-1]
    top_terms = [feature_names[j] for j in top_term_indices]
    print(f"  Topic {i+1}: {top_terms}")
```
