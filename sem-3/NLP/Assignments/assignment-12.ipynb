```python
# Experiment 12
import gensim
from gensim import corpora
from nltk.tokenize import word_tokenize
import nltk

# Ensure NLTK data is available
try:
    nltk.data.find('tokenizers/punkt')
except:
    nltk.download('punkt', quiet=True)

# 1. Sample Corpus
documents = [
    "Federal reserve announces rate hike amidst inflation concerns.",
    "The new electric car has a long range and sleek design.",
    "Tesla introduces novel battery technology for vehicles.",
    "Central bank meeting discusses monetary policy and job growth.",
    "Investors monitor bond yields and commodity prices."
]

# 2. Preprocessing: Tokenization and creating a dictionary
tokenized_docs = [word_tokenize(doc.lower()) for doc in documents]
dictionary = corpora.Dictionary(tokenized_docs)
corpus_bow = [dictionary.doc2bow(doc) for doc in tokenized_docs]

print(f"Number of Unique Tokens: {len(dictionary)}")
print(f"Sample Document 1 (BoW):\n{corpus_bow[0]}\n")

# 3. Train the LDA Model
num_topics = 2 # We expect one topic for 'Finance' and one for 'Cars'
lda_model = gensim.models.LdaModel(
    corpus=corpus_bow,
    id2word=dictionary,
    num_topics=num_topics,
    random_state=42,
    passes=10 # Number of training passes
)

# 4. Output Results (Topics and associated words)
print("-" * 50)
print(f"LDA Topic Modeling (K={num_topics} Topics):")
topics = lda_model.print_topics(num_words=5)
for topic_id, word_dist in topics:
    print(f"Topic {topic_id + 1}: {word_dist}")

# 5. Document-Topic Distribution (Optional)
print("\nDocument Topic Distribution (Score for Topic 1, Topic 2):")
for i, doc_bow in enumerate(corpus_bow):
    doc_topics = lda_model.get_document_topics(doc_bow, minimum_probability=0.1)
    topic_scores = {topic_id: score for topic_id, score in doc_topics}
    score1 = topic_scores.get(0, 0.0)
    score2 = topic_scores.get(1, 0.0)
    print(f"Doc {i+1}: ({score1:.3f}, {score2:.3f})")
```
