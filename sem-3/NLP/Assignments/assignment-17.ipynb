```python
# Experiment 17
from gtts import gTTS
import os
from io import BytesIO

# 1. Define the text to synthesize
text_to_read = "Hello, this is a demonstration of Text to Speech synthesis using the gTTS library. It saves the audio as an MP3 file."
output_filename = "tts_output.mp3"

print("--- Text-to-Speech (TTS) Synthesis Demo ---")
print(f"Text: '{text_to_read}'")

# 2. Create the gTTS object
try:
    tts = gTTS(text=text_to_read, lang='en', slow=False)
    
    # 3. Save the generated audio (MP3 format)
    # Use BytesIO to simulate file generation in memory
    # In a real local environment, use: tts.save(output_filename)
    
    audio_fp = BytesIO()
    tts.write_to_fp(audio_fp)
    audio_size = audio_fp.tell()
    
    print(f"\nSuccessfully synthesized audio.")
    print(f"The audio data was generated (simulated size: {audio_size} bytes).")
    print(f"If running locally, this would be saved as '{output_filename}' for playback.")

except Exception as e:
    print(f"\nFailed to generate TTS audio. Error: {e}")
    print("Ensure you have an active internet connection for gTTS to work.")
```
