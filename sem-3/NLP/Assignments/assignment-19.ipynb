```python
# Experiment 19
from transformers import pipeline

# 1. Initialize the Sentiment Analysis pipeline
# We use a model fine-tuned for sentiment analysis
try:
    sentiment_pipeline = pipeline(
        "sentiment-analysis",
        model="distilbert-base-uncased-finetuned-sst-2-english"
    )
except Exception as e:
    print(f"Error loading model.")
    exit()

# 2. Define sample sentences
sentences = [
    "I absolutely loved the simplicity of this transformer code.",
    "The weather today is terrible and I feel sad.",
    "This is an incredibly neutral statement.",
    "The latency was a major drawback of the system.",
]

print("--- Sentiment Analysis using Pre-trained BERT (DistilBERT) ---")
print(f"Model used: distilbert-base-uncased-finetuned-sst-2-english\n")

# 3. Run analysis and output results
print(f"{'Sentence':<50} | {'Sentiment':<10} | {'Score':<5}")
print("-" * 70)

results = sentiment_pipeline(sentences)

for sentence, result in zip(sentences, results):
    label = result['label']
    score = result['score']
    
    print(f"{sentence[:48]+'...':<50} | {label:<10} | {score:.3f}")
```
