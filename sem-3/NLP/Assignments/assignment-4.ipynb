```python
## Experiment 4
import spacy

# Load the small English model
try:
    nlp = spacy.load("en_core_web_sm")
except OSError:
    print("Error: spaCy model 'en_core_web_sm' not found.")
    print("Please run: python -m spacy download en_core_web_sm")
    exit()

# 1. Sample text containing various entities
text = (
    "Apple is set to acquire a startup in London for $1 billion. "
    "Tim Cook announced the deal on Tuesday, October 26, 2024. "
    "The new office will employ over 500 people."
)
print(f"Original Text:\n{text}\n")

# 2. Process the text with the spaCy model
doc = nlp(text)

# 3. Extract and classify named entities
print("Named Entities Found:")
print("-" * 35)
for ent in doc.ents:
    # ent.text: The recognized entity text
    # ent.label_: The label/type of the entity (e.g., ORG, GPE, MONEY, DATE)
    print(f"  - Entity: {ent.text:<20} | Type: {ent.label_}")
```
