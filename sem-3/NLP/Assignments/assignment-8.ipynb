```python
# Experiment 8
import numpy as np
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.svm import SVC
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score

# 1. Create a simple toy dataset (Sentiment: 0 = Negative, 1 = Positive)
texts = [
    "The movie was incredibly boring and slow.",  # Negative (0)
    "I absolutely loved the acting and plot!",    # Positive (1)
    "A waste of time, I would not recommend it.", # Negative (0)
    "This is the best film I have ever seen.",    # Positive (1)
    "It was okay, but the ending was poor.",      # Negative (0)
    "I felt energized and happy after watching.", # Positive (1)
]
labels = np.array([0, 1, 0, 1, 0, 1])

# 2. Split data
X_train, X_test, y_train, y_test = train_test_split(
    texts, labels, test_size=0.33, random_state=42
)
print(f"Training Samples: {len(X_train)}, Test Samples: {len(X_test)}")

# 3. Feature Extraction (TF-IDF)
tfidf_vectorizer = TfidfVectorizer(stop_words='english')
X_train_vectorized = tfidf_vectorizer.fit_transform(X_train)
X_test_vectorized = tfidf_vectorizer.transform(X_test)

# 4. Train the Support Vector Machine (SVC) model
# Using a linear kernel often works well for text classification
svm_model = SVC(kernel='linear', random_state=42)
svm_model.fit(X_train_vectorized, y_train)

# 5. Predict on the test set
y_pred = svm_model.predict(X_test_vectorized)

# 6. Evaluate the model
accuracy = accuracy_score(y_test, y_pred)
print("-" * 50)
print("SVM Classifier Performance:")
print(f"True Labels (Test): {y_test}")
print(f"Predicted Labels:   {y_pred}")
print(f"Accuracy: {accuracy:.2f}")

# 7. Test on a new text
new_text = ["Simply stunning and perfectly executed film."]
new_text_vectorized = tfidf_vectorizer.transform(new_text)
prediction = svm_model.predict(new_text_vectorized)[0]
sentiment = 'Positive' if prediction == 1 else 'Negative'
print(f"\nPrediction for new text: {sentiment}")
```
