```python
# Experiment 3
import nltk
from nltk.stem import PorterStemmer, WordNetLemmatizer

# Ensure WordNet data is available
try:
    nltk.data.find('corpora/wordnet')
except:
    nltk.download('wordnet', quiet=True)
try:
    nltk.data.find('corpora/omw-1.4')
except:
    nltk.download('omw-1.4', quiet=True)

# 1. Initialize Stemmer and Lemmatizer
stemmer = PorterStemmer()
lemmatizer = WordNetLemmatizer()

# 2. Sample words
words = ["running", "cats", "studies", "better", "corpora", "saw"]

print(f"Original Words: {words}\n")
print("-" * 35)
print(f"{'Word':<10} | {'Stem (Porter)':<15} | {'Lemma (Noun)':<10}")
print("-" * 35)

# 3. Compare outputs
for word in words:
    stem_output = stemmer.stem(word)
    # Lemmatizer requires Part-of-Speech tag for accuracy. Default is 'n' (Noun).
    lemma_output_n = lemmatizer.lemmatize(word, pos='n')
    
    print(f"{word:<10} | {stem_output:<15} | {lemma_output_n:<10}")

# Example demonstrating POS sensitivity for Lemmatization
print("\n--- POS Sensitivity for 'running' ---")
print(f"Running (Noun):     {lemmatizer.lemmatize('running', pos='n')}")
print(f"Running (Verb, 'v'): {lemmatizer.lemmatize('running', pos='v')}")
```
