```python
# Experiment 20
from transformers import pipeline

# 1. Initialize the Question Answering pipeline
# The pipeline uses a model fine-tuned on the SQuAD dataset
try:
    qa_pipeline = pipeline(
        "question-answering",
        model="distilbert-base-cased-distilled-squad"
    )
except Exception as e:
    print(f"Error loading model.")
    exit()

# 2. Define the context (paragraph) and the question
context = (
    "The invention of the printing press by Johannes Gutenberg around 1440 "
    "is widely regarded as the most important event of the modern period. "
    "It revolutionized information dissemination, making books much cheaper "
    "and more accessible to the general public. Before its invention, books "
    "had to be copied by hand, a slow and expensive process."
)
question = "Who invented the printing press and when?"

print("--- Question Answering (QA) using BERT (DistilBERT) ---")
print("-" * 70)
print(f"Context:\n{context}\n")
print(f"Question: {question}")
print("-" * 70)

# 3. Run the QA system
result = qa_pipeline(
    question=question,
    context=context
)

# 4. Output the result
answer = result['answer']
score = result['score']
start_index = result['start']
end_index = result['end']

print("QA Result:")
print(f"  Answer: {answer}")
print(f"  Confidence Score: {score:.4f}")
```
