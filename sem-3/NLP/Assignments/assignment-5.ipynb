```python
# Experiment 5
import nltk
from nltk.parse import RecursiveDescentParser

# Ensure NLTK data is available for pretty printing
try:
    nltk.data.find('tokenizers/punkt')
except:
    nltk.download('punkt', quiet=True)

# 1. Define a Context-Free Grammar (CFG)
grammar_definition = """
    S -> NP VP
    NP -> Det N | N | Det N PP
    VP -> V NP PP
    NP -> Det N PP
    PP -> P NP
    Det -> 'the' | 'a'
    N -> 'dog' | 'park' | 'man' | 'telescope'
    V -> 'saw'
    P -> 'with' | 'in'
"""
cfg = nltk.CFG.fromstring(grammar_definition)

# 2. Initialize the parser (Recursive Descent is one simple type)
parser = RecursiveDescentParser(cfg)

# 3. Define a complex sentence (with ambiguity)
sentence = ['the', 'man', 'saw', 'the', 'dog', 'with', 'a', 'telescope']
print(f"Sentence: {' '.join(sentence)}\n")

# 4. Parse the sentence and generate tree objects
print("Generated Parse Trees:")
print("-" * 60)
trees = list(parser.parse(sentence))

# 5. Visualize (print) the parse trees
if trees:
    for i, tree in enumerate(trees):
        print(f"Tree {i+1}:")
        # Use pretty_print() to visualize the structure
        tree.pretty_print()
else:
    print("No parse found for the sentence based on the grammar.")
```
