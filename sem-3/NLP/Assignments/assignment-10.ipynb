```python
# Experiment 10
import numpy as np

# --- Simplified Viterbi Implementation ---
# States: S1=H (High), S2=L (Low) - Hidden States
# Observations: O1=U (Up), O2=D (Down) - Visible Observations

# 1. Define Model Parameters
states = ('H', 'L')
observations = ('U', 'D', 'U', 'D') # Sequence
S_idx = {'H': 0, 'L': 1}
O_idx = {'U': 0, 'D': 1}

# Initial Probabilities (Pi)
start_prob = np.array([0.8, 0.2]) # P(H), P(L)

# Transition Matrix (A): A[i, j] = P(State j | State i)
trans_prob = np.array([[0.7, 0.3], [0.4, 0.6]])

# Emission Matrix (B): B[i, j] = P(Obs j | State i)
emission_prob = np.array([[0.9, 0.1], [0.2, 0.8]])

def viterbi(obs_seq, states, start_p, trans_p, emit_p):
    T = len(obs_seq) # Length of observation sequence
    N = len(states)  # Number of states
    
    # 1. Initialization: Create Viterbi and Backpointer tables
    V = np.zeros((N, T))
    path = np.zeros((N, T), dtype=int)
    
    # Time t=0 (First observation)
    for i in range(N):
        obs_idx = O_idx[obs_seq[0]]
        V[i, 0] = start_p[i] * emit_p[i, obs_idx]
    
    # 2. Recursion: For t > 0
    for t in range(1, T):
        obs_idx = O_idx[obs_seq[t]]
        for j in range(N): # Current state
            likelihoods = V[:, t-1] * trans_p[:, j] * emit_p[j, obs_idx]
            V[j, t] = np.max(likelihoods)
            path[j, t] = np.argmax(likelihoods)
    
    # 3. Termination and Backtracking
    most_likely_last_state = np.argmax(V[:, T-1])
    best_path = [most_likely_last_state]
    current_state = most_likely_last_state
    
    for t in range(T - 1, 0, -1):
        current_state = path[current_state, t]
        best_path.insert(0, current_state)
        
    result_path = [states[i] for i in best_path]
    max_prob = np.max(V[:, T-1])
    return result_path, max_prob, V

# 2. Run the Viterbi algorithm
predicted_sequence, probability, V_table = viterbi(
    observations, states, start_prob, trans_prob, emission_prob
)

# 3. Output results
print("Viterbi Algorithm Implementation (Custom)")
print("-" * 50)
print(f"Observations (Price Movement): {observations}")
print(f"Predicted Hidden States (Volatility): {predicted_sequence}")
print(f"Probability of Sequence: {probability:.4f}")
print("\nPartial Viterbi Table (V):")
print(V_table.round(5))
```
