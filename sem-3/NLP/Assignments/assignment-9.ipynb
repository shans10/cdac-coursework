```python
# Experiment 9
import numpy as np
from hmmlearn import hmm

# --- HMM Model for Weather (Hidden State) and Activity (Observation) ---
# States: S1=Rainy, S2=Sunny (Hidden)
# Observations: O1=Walk, O2=Shop, O3=Clean (Visible)

# 1. Define the model parameters manually (for simplicity)

# State-Transition Probabilities: P(St | St-1)
trans_prob = np.array([[0.7, 0.3], [0.4, 0.6]])

# Emission Probabilities: P(Ot | St)
emission_prob = np.array([[0.1, 0.6, 0.3], [0.5, 0.2, 0.3]])

# Initial State Probabilities: P(S1), P(S2)
start_prob = np.array([0.6, 0.4])

# Mapping for clarity
states_map = {0: "Rainy", 1: "Sunny"}
observations_map = {0: "Walk", 1: "Shop", 2: "Clean"}

# 2. Sequence of Observations (Activities over several days)
# Clean (2), Walk (0), Shop (1), Clean (2), Walk (0)
observations = np.array([[2, 0, 1, 2, 0]]).T  # Needs to be a column vector

# 3. Initialize the HMM
model = hmm.CategoricalHMM(n_components=len(states_map), random_state=42, n_iter=100)
model.startprob_ = start_prob
model.transmat_ = trans_prob
model.emissionprob_ = emission_prob

# 4. Predict the most probable sequence of hidden states (Weather)
# This uses the Viterbi algorithm internally
logprob, hidden_states = model.decode(observations, algorithm="viterbi")

print("HMM for Weather (Hidden) and Activity (Observed)")
print("-" * 50)
print(f"Observation Sequence: {[observations_map[o[0]] for o in observations]}")

# Map the predicted state indices back to names
predicted_states = [states_map[s] for s in hidden_states]

print(f"\nMost Probable Hidden State Sequence (Weather):")
print(f"{'Time Step':<10} | {'Activity':<10} | {'Predicted Weather':<20}")
print("-" * 50)
for i, (obs, state) in enumerate(zip(observations, predicted_states)):
    print(f"{i+1:<10} | {observations_map[obs[0]]:<10} | {state:<20}")
```
